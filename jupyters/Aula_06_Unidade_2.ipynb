# 2. Ausência estrutural e Imputação
Agora que encontramos colunas com faltantes, precisamos analisar as condições de ausencia do valor, antes de escolher qualquer tecnica de imputação.
Na base de imóveis, **floor** e **total_floors** estão ausentes nas casas. Não houve falha de cadastro. Esses atributos descrevem a posição de uma unidade em um edifício e, portanto, não se aplicam a casas. Esse tipo de lacuna é chamado de ausência estrutural, pois decorre da definição da variável para determinado tipo de observação. Imputar um número produziria um valor sem referente real.

```python
andar_por_tipo = (
                imoveis_df
                .assign(sem_andar = imoveis_df["andar"].isna()) # adiciona uma coluna de filtro
                .groupby("tipo_imovel")
                .agg(
                    quantidade=("tipo_imovel","count"),
                    proporcao_sem_andar = ("sem_andar","mean") # utiliza a coluna de filtro pra fazer a media
                    )
                )
andar_por_tipo.head()
```

Corforme apontado e observado, os ausentes na realidade guardam uma informação importante sobre como o dataset distingue entre casas e apartamento

## 2.1 Sobre a imputação de valores ausentes

A imputação substitui valores ausentes por estimativas, usando estatísticas da própria coluna ou outras variáveis. Mesmo métodos simples podem concentrar artificialmente valores e reduzir a variabilidade dos dados.

Como não conhecemos os valores verdadeiros das lacunas reais, não é possível calcular diretamente o erro da imputação. Para avaliar os métodos, podemos simular esse cenário apagando valores de uma base completa, imputando-os e comparando os resultados com os valores originais.

Primeiro, implementamos uma função para apagar uma fração dos valores, mantendo a base completa como referência.


```python
# Função de apagamento e base-gabarito
rng = np.random.default_rng(42)

def erase_at_random(values, share):
    return values.mask(rng.random(len(values)) < share)

imoveis_true = (
    imoveis_df
    .dropna(subset=["condominio_mensal", "vagas_garagem"])
    .reset_index(drop=True)
)
imoveis_true.head()
```

Em seguida, criarei lacunas em três variáveis de tipos diferentes: discreta, contínua e categórica.

```python
# Apaga valores em três colunas de tipos diferentes
imoveis_missing = (
    imoveis_true
    .assign(
        quartos=lambda df: erase_at_random(df["quartos"], .10),
        aluguel_mensal=lambda df: erase_at_random(df["aluguel_mensal"], .15),
        mobiliado=lambda df: erase_at_random(df["mobiliado"], .08),
    )
)
```

Conferimos quantos valores foram apagados:

```python
tabela_proporção_ausentes = imoveis_missing[["quartos","aluguel_mensal","mobiliado"]].isna().sum().to_frame("ausentes")

tabela_proporção_ausentes["proporcao"] = (tabela_proporção_ausentes["ausentes"]/len(imoveis_missing)).round(2)

tabela_proporção_ausentes.head()
```

## 2.2 Média, mediana e mediana agrupada
As estratégias mais simples substituem todas as lacunas por uma única constante calculada com os valores observados da coluna. 
A média utiliza toda a magnitude dos dados e, por isso, responde fortemente a valores extremos. 
A mediana depende apenas da ordenação e é mais resistente a caudas longas.


```python
# Constantes para a imputação simples
rent_mean = (
    imoveis_missing.aluguel_mensal
    .mean()
)

rent_median = (
    imoveis_missing.aluguel_mensal
    .median()
)
```

A mediana por `tier` considera o padrão do imóvel para estimar os valores ausentes. Caso um grupo não tenha valores observados, utiliza-se a mediana global como alternativa.


```python
# Imputação pela mediana de cada tier
tier_based_median = (
                    imoveis_missing.
                    groupby("tier").
                    agg(mediana_por_tier = ("aluguel_mensal","median"))
                    )

mediana_da_linha = imoveis_missing["tier"].map(tier_based_median["mediana_por_tier"]) # coluna da mediana por tier de todas as tuplas

imoveis_input_por_tier = imoveis_missing.assign(aluguel_mensal=imoveis_missing["aluguel_mensal"].fillna(mediana_da_linha))

imoveis_input_por_tier.head(10)
```

Reunimos as versões do aluguel para compará-las lado a lado:

```python
# Versões do aluguel a comparar
rent_versions = {
    "Original": imoveis_true.aluguel_mensal,
    "Com lacunas": imoveis_missing.aluguel_mensal,
    "Média": imoveis_missing.aluguel_mensal.fillna(rent_mean),
    "Mediana": imoveis_missing.aluguel_mensal.fillna(rent_median),
    "Mediana por tier": imoveis_input_por_tier.aluguel_mensal,
}

# Histograma de cada versão do aluguel
fig, axes = plt.subplots(1, 5, figsize=(20, 4), sharey=True)

limit = imoveis_true.aluguel_mensal.quantile(.99)

for ax, (label, values) in zip(axes, rent_versions.items()):
    sns.histplot(
        values.dropna(),
        bins=50,
        binrange=(0, limit),
        ax=ax,
    )
    ax.set(
        title=label,
        xlabel="Aluguel (R$)",
        ylabel="Imóveis",
    )
    if label == "Média":
        ax.axvline(
            rent_mean,
            color="#C44E52",
            linestyle="--",
        )
    if label == "Mediana":
        ax.axvline(
            rent_median,
            color="#C44E52",
            linestyle="--",
        )

plt.tight_layout()
plt.show()
```

## 2.3 Erro absoluto médio (MAE)

Os histogramas mostram como a imputação altera a distribuição, mas não medem a precisão dos valores estimados. Como os valores apagados são conhecidos na base original, podemos comparar as estimativas com os valores verdadeiros.

O **MAE (*Mean Absolute Error*)** calcula a média dos erros absolutos, evitando que erros positivos e negativos se cancelem.

$$
MAE = \frac{1}{n}\sum_{i=1}^{n}|y_i-\hat{y}_i|
$$

O resultado fica na mesma unidade da variável. Por exemplo, um MAE de R$ 500 indica que as estimativas erram, em média, R$ 500 em relação aos valores reais.
* **Considerar apenas os valores imputados** no cálculo do MAE; Afinal, incluir os valores observados reduziria artificialmente o erro.
* **A métrica favorece estratégias diferentes:** a mediana minimiza os erros absolutos, enquanto a média minimiza os erros quadráticos. Portanto, espera-se um MAE menor com a mediana, especialmente em distribuições assimétricas, como a dos aluguéis.

Primeiro, recuperamos no gabarito os valores verdadeiros das posições apagadas:


```python
# Posições apagadas e seus valores verdadeiros
alugueis_faltantes = imoveis_missing.aluguel_mensal.isna()
aluguel_medio = rent_mean
aluguel_mediana = rent_median
aluguel_verdadeiro = imoveis_true.loc[alugueis_faltantes, "aluguel_mensal"]
```

```python
# MAE de cada estratégia, apenas nas posições imputadas
(
    pd.Series(
        {
            "Média": (
                (aluguel_verdadeiro - aluguel_medio)
                .abs()
                .mean()
            ),
            "Mediana": (
                (aluguel_verdadeiro - aluguel_mediana)
                .abs()
                .mean()
            ),
            "Mediana por tier": (
                (aluguel_verdadeiro - imoveis_input_por_tier.loc[alugueis_faltantes, "aluguel_mensal"])
                .abs()
                .mean()
            ),
        },
        name="MAE (R$)",
    )
    .round(2)
    .to_frame()
)
```

Quanto menor o MAE, menor o erro médio das imputações. A mediana apresenta melhor resultado que a média, e a mediana por `tier` reduz ainda mais o erro ao considerar o padrão do imóvel. Essa avaliação vale para o cenário MCAR simulado, não garantindo o mesmo desempenho em dados com outros padrões de ausência.


## 2.4 Moda e variáveis discretas
Para variáveis categóricas nominais, média e mediana não se aplicam. Uma alternativa é imputar a moda, ou seja, a categoria mais frequente.


```python
moda_mobilia =imoveis_missing.mobiliado.mode().iloc[0] #pega a primeira moda encontrada, porque moda retorna uma lista de 1 ou mais modas
```

Depois da imputação, todas as lacunas passam a contar como ocorrências dessa categoria. A frequência da moda aumenta artificialmente, um efeito que precisa ser considerado ao interpretar a distribuição resultante.

```python
count_pre_input = imoveis_missing.mobiliado.value_counts(dropna=False)
count_pos_input = imoveis_missing.mobiliado.fillna(moda_mobilia).value_counts()

pd.concat({"Antes": count_pre_input,"Imputacao por moda": count_pos_input},axis=1).fillna(0)
```

Em uma variável quantitativa discreta, como o número de quartos, a média é matematicamente definida, mas pode produzir um valor incompatível com o domínio da variável, como uma quantidade fracionária de quartos:

```python
# Média, mediana e moda do número de quartos

pd.Series({"média": imoveis_missing.quartos.mean(),"mediana": imoveis_missing.quartos.median(),"moda": (imoveis_missing.quartos.mode().iloc[0])},name="quartos").to_frame().round(2)
```

## 2.5 Resumo da imputação

- **Contínua e simétrica:** média.
- **Assimétrica ou com extremos:** mediana.
- **Categórica ou discreta:** moda ou mediana.
- **Varia entre grupos conhecidos:** estatística por grupo (mediana por tier).

**Importante:** crie uma coluna indicadora para distinguir valores observados dos imputados.

Após a imputação, o próximo passo é o **escalonamento**, que ajusta a representação numérica das variáveis, e as **transformações não lineares**, que alteram a forma da distribuição.

# 3. Escalonamento Linear
O **escalonamento** aplica transformações do tipo \(x'=(x-c)/s\), em que \(c\) é uma medida de centro e \(s\) uma medida de escala. Preserva a ordenação, a assimetria e a curtose, alterando apenas a unidade da variável.

Seu objetivo é colocar variáveis numéricas em escalas comparáveis, evitando que atributos com valores maiores dominem os cálculos. É importante em algoritmos baseados em distâncias, como k-NN e k-means, e em modelos com gradiente e regularização.

As técnicas mais comuns são o **z-score**, que utiliza média e desvio-padrão, e o **Min-Max**, que utiliza mínimo e amplitude. Já o `Normalizer` do scikit-learn normaliza cada amostra pela sua norma vetorial.

## 3.1 Três escaladores

Os três escaladores diferem na escolha do centro e da escala, e essa escolha determina a sensibilidade a valores extremos.

| Técnica | Centro e escala | Sensibilidade a extremos | Indicação |
|---|---|---|---|
| `StandardScaler` | Média e desvio-padrão | Alta | Sem extremos importantes |
| `RobustScaler` | Mediana e IQR | Baixa | Com valores extremos |
| `MinMaxScaler` | Mínimo e máximo | Muito alta | Limites estáveis |


Aplicamos os três escaladores ao aluguel.


Os escaladores do scikit-learn recebem um DataFrame e devolvem uma matriz. .ravel() converte essa matriz em vetor.

```python
#aplicando o scalers no alguel

aluguel_df = imoveis_true[["aluguel_mensal"]]

scaled ={
"Original": aluguel_df.aluguel_mensal.to_numpy(),

"StandardScaler": StandardScaler().fit_transform(aluguel_df).ravel(),

"RobustScaler" : RobustScaler().fit_transform(aluguel_df).ravel(),

"MinMaxScaler" : MinMaxScaler().fit_transform(aluguel_df).ravel()
}
```

`[[ ]]` mantém os dados em formato 2D. `.fit_transform()` estima os parâmetros do escalador e os aplica aos dados.

```python
# Histograma do aluguel em cada escala
fig, axes = plt.subplots(1, 4, figsize=(16, 4), sharey=True)

for ax, (label, values) in zip(axes, scaled.items()):
    sns.histplot(values, bins=50, ax=ax)
    ax.set(
        title=f"{label}\nassimetria = {stats.skew(values):.2f}",
        xlabel="Escala",
        ylabel="Imóveis",
    )

plt.tight_layout()
plt.show()
```

Os quatro histogramas têm a mesma forma e a mesma assimetria. Apenas os valores do eixo horizontal mudam, porque cada método define uma nova origem e uma nova unidade de medida.

## 3.2 Sensibilidade a extremos
O Min-Max usa o mínimo e o máximo observados como referência. Um único aluguel extremo aumenta a amplitude usada no denominador e comprime os demais valores em uma faixa estreita próxima de zero.

Para ver o efeito, acrescentamos um aluguel fictício de R\$ 500 mil e o descartamos depois do ajuste:

```python
# min-max ajustado com um aluguel fictício de R$ 500 mil
com_extremo = pd.concat(
    [aluguel_df, pd.DataFrame({"aluguel_mensal": [500_000.]})],
    ignore_index=True,
)

minmax_extremo = MinMaxScaler().fit_transform(com_extremo).ravel()[:-1] # descarta o valor fictício
```

```python
# min-max sem e com o valor extremo
minmax_versions = {
    "Sem extremo": scaled["MinMaxScaler"],
    "Com aluguel de R$ 500 mil": minmax_extremo,
}

fig, axes = plt.subplots(1, 2, figsize=(10, 4), sharey=True)

for ax, (label, values) in zip(axes, minmax_versions.items()):
    sns.histplot(values, bins=50, binrange=(0, 1), ax=ax)
    ax.set(
        title=label,
        xlabel="Escala Min-Max",
        ylabel="Imóveis",
    )

plt.tight_layout()
plt.show()
```

Com o extremo, quase todos os imóveis ficam concentrados no início da escala. O `RobustScaler`, baseado na mediana e no IQR, é muito menos sensível a esse efeito: o ponto extremo continua distante depois da transformação, mas não determina o centro nem a escala usados para a maioria das observações.

**Importante:** escalonar muda a régua, não a forma.

O aluguel continua com a mesma cauda longa, apenas em outra unidade.

# 4. Transformações não lineares
Aqui, normalizar significa transformar a própria distribuição, em geral para torná-la mais simétrica e mais próxima da normal, e não apenas mudar a origem e a unidade como no escalonamento.

Uma função não linear comprime regiões diferentes do eixo em intensidades diferentes. O logaritmo, por exemplo, reduz mais as distâncias entre valores altos do que entre valores baixos: na escala log, a distância entre R\$ 1.000 e R\$ 2.000 é a mesma que entre R\$ 10.000 e R\$ 20.000, pois os dois pares representam a mesma razão. A ordenação é preservada, mas a assimetria, a curtose e a influência dos extremos mudam.

O custo é a interpretação, já que os valores na escala transformada deixam de estar em reais. Além disso, a escolha da transformação depende dos valores que a variável aceita: algumas exigem valores estritamente positivos, outras aceitam zero ou negativos.



## 4.1 Aluguel positivo: log e Box-Cox
Todos os aluguéis da base são estritamente positivos, condição necessária para aplicar `log` e Box-Cox. Comparamos o logaritmo, adequado a valores monetários com variação multiplicativa, com o Box-Cox, que estima um parâmetro de potência (λ) a partir dos dados.

```python
# aluguel original, em log e em Box-Cox
aluguel = imoveis_true["aluguel_mensal"]

aluguel_boxcox, lambda_boxcox = stats.boxcox(aluguel)

aluguel_transformado = {
    "Original": aluguel,
    "log": np.log(aluguel),
    f"Box-Cox (λ={lambda_boxcox:.2f})": aluguel_boxcox,
}
```

```python
# histograma de cada transformação do aluguel
fig, axes = plt.subplots(1, 3, figsize=(14, 4), sharey=True)

for ax, (label, values) in zip(axes, aluguel_transformado.items()):
    sns.histplot(values, bins=50, ax=ax)
    ax.set(
        title=f"{label}\nassimetria = {stats.skew(values):.2f}",
        xlabel="Escala",
        ylabel="Imóveis",
    )

plt.tight_layout()
plt.show()
```

O Box-Cox estima o λ por máxima verossimilhança, assumindo que os valores transformados se aproximam de uma normal. Quando λ = 0, ele equivale ao logaritmo natural. Como o λ estimado ficou próximo de zero, a transformação é muito parecida com o log, o que sustenta o uso do `log` para o aluguel.

Nem toda variável é estritamente positiva. Quando há zeros, o logaritmo deixa de ser definido e é preciso recorrer a variantes.

## 4.2 Distância com zeros: `log1p` e Yeo-Johnson
A distância (`distancia_centro_minutos`) pode ser igual a zero, então `log` e Box-Cox não se aplicam diretamente. Duas alternativas são o `log1p`, que calcula $\log(1+x)$, e o Yeo-Johnson, uma transformação de potência definida também para zero e valores negativos.

Com `standardize=False`, o `PowerTransformer` apenas transforma, sem escalonar.

```python
# distância original, em log1p e em Yeo-Johnson
distancia = imoveis_true["distancia_centro_minutos"]

power_transformer = PowerTransformer(method="yeo-johnson", standardize=False)
distancia_yj = power_transformer.fit_transform(imoveis_true[["distancia_centro_minutos"]]).ravel()

distancia_transformada = {
    "Original": distancia,
    "log1p": np.log1p(distancia),
    f"Yeo-Johnson (λ={power_transformer.lambdas_[0]:.2f})": distancia_yj,
}
```

```python
# histograma de cada transformação da distância
fig, axes = plt.subplots(1, 3, figsize=(14, 4), sharey=True)

for ax, (label, values) in zip(axes, distancia_transformada.items()):
    sns.histplot(values, bins=50, ax=ax)
    ax.set(
        title=f"{label}\nassimetria = {stats.skew(values):.2f}",
        xlabel="Escala",
        ylabel="Imóveis",
    )

plt.tight_layout()
plt.show()
```

Comparando as assimetrias:

- `log1p` aceita zeros, mas aqui comprime a cauda direita além do necessário;
- Yeo-Johnson aceita qualquer sinal e estima uma transformação mais branda.

## 4.3 Resumo das transformações

| Técnica | Valores aceitos | Interpretação |
|---|---|---|
| `log` | Positivos | Boa: diferenças representam razões |
| `log1p` | Zero e positivos | Boa, com inversa `expm1` |
| Box-Cox | Positivos | Menos direta |
| Yeo-Johnson | Qualquer sinal | Menos direta |

A transformação de Box-Cox é dada por:

$$
x^{(\lambda)} =
\begin{cases}
\dfrac{x^\lambda - 1}{\lambda}, & \lambda \ne 0 \\[6pt]
\log(x), & \lambda = 0
\end{cases}
$$

Para voltar à unidade original, use `np.exp` após o `log` e `np.expm1` após o `log1p`.

**Importante:** transformar muda a forma. A redução da assimetria deve ser avaliada junto com a perda de interpretação direta na unidade original.

# 5. Aplicação e Analise Final
Até aqui, cada técnica foi estudada isoladamente. Agora reunimos as decisões em um único fluxo, aplicado à base completa (`imoveis_df`) com suas lacunas reais. A ordem importa: as colunas indicadoras vêm antes da imputação, enquanto as posições ausentes ainda são identificáveis, e a imputação vem antes das transformações, para que as medianas sejam calculadas na unidade original.



## 5.1 Registrar as lacunas
As colunas indicadoras são criadas **antes** da imputação. Depois dela, o rastro se perderia.

```python
imoveis_processado = imoveis_df.copy()

imoveis_processado["condominio_ausente"] = imoveis_processado["condominio_mensal"].isna()
imoveis_processado["vagas_ausente"] = imoveis_processado["vagas_garagem"].isna()
```

## 5.2 Imputar pela mediana do grupo
Os grupos combinam `tipo_imovel` e `tier`. Grupos sem observações recebem a mediana global.

```python
grupos = ["tipo_imovel", "tier"]

condominio_mediana_grupo = imoveis_processado.groupby(grupos)["condominio_mensal"].transform("median")
vagas_mediana_grupo = imoveis_processado.groupby(grupos)["vagas_garagem"].transform("median")

imoveis_processado["condominio_mensal"] = (
    imoveis_processado["condominio_mensal"]
    .fillna(condominio_mediana_grupo)
    .fillna(imoveis_df["condominio_mensal"].median())
)

imoveis_processado["vagas_garagem"] = (
    imoveis_processado["vagas_garagem"]
    .fillna(vagas_mediana_grupo)
    .fillna(imoveis_df["vagas_garagem"].median())
)
```

Conferimos que não restaram lacunas e quantos valores foram estimados:

```python
restantes = imoveis_processado[["condominio_mensal", "vagas_garagem"]].isna().sum()
estimados = imoveis_processado[["condominio_ausente", "vagas_ausente"]].sum()

print("\n Lacunas restantes :")
print(restantes)

print("\n Valores estimados :")
print(estimados)
```

## 5.3 Reduzir as caudas
Seguimos o resumo da seção 4.3:

- área e valores monetários são positivos e recebem `log`;
- a distância, que pode ser zero, recebe Yeo-Johnson.

```python
for coluna in ["area", "aluguel_mensal", "iptu_mensal", "total"]:
    imoveis_processado[coluna] = np.log(imoveis_processado[coluna])

imoveis_processado["distancia_centro_minutos"] = stats.yeojohnson(imoveis_processado["distancia_centro_minutos"])[0]
```

## 5.4 Antes e depois
Comparamos as distribuições das cinco variáveis transformadas:

```python
colunas = ["area", "aluguel_mensal", "iptu_mensal", "total", "distancia_centro_minutos"]
rotulos = ["Área", "Aluguel", "IPTU", "Total", "Distância"]
```

```python
fig, axes = plt.subplots(2, 5, figsize=(18, 7))

for coluna, rotulo, ax_antes, ax_depois in zip(colunas, rotulos, axes[0], axes[1]):
    antes = imoveis_df[coluna]
    depois = imoveis_processado[coluna]

    sns.histplot(antes, bins=50, ax=ax_antes, color="lightblue")
    ax_antes.set(
        title=f"{rotulo}: antes\nassimetria = {stats.skew(antes):.2f}",
        xlabel="",
        ylabel="Imóveis",
    )

    sns.histplot(depois, bins=50, ax=ax_depois, color="#91D6A6")
    ax_depois.set(
        title=f"{rotulo}: depois\nassimetria = {stats.skew(depois):.2f}",
        xlabel="",
        ylabel="Imóveis",
    )

plt.tight_layout()
plt.show()
```

Na linha de baixo, as distribuições apresentam menor assimetria e os valores extremos ocupam uma parcela menor do eixo. Isso não significa que os extremos foram removidos, apenas que suas distâncias em relação ao centro foram comprimidas pela transformação.

**Evite vazamento:** em um projeto de modelagem, medianas, parâmetros λ e parâmetros dos escaladores devem ser estimados **apenas no conjunto de treino** e depois aplicados ao de teste. Ajustá-los antes da separação deixa informações do teste influenciarem o pré-processamento.

## Perguntas para consolidar

#### Pergunta 1: imputar sempre melhora a análise?

A seção 2 mostrou lacunas que não deveriam ser preenchidas. Em que situação uma estratégia com MAE baixo no experimento controlado ainda pode piorar a análise quando aplicada à base real?

### Resposta :
Alguns datasets possuem lacunas e ausentes que são estruturais, de modo que fazem parte da composição dos dados e carregam algum significado, esse pode ser inferido a partir de outras colunas como no exemplo "andar" que se relaciona diretamente com o tipo de imovel. Nesses casos, preencher a lacuna criaria um valor que não existe na realidade e comprometeria a logica do dataset. Metricas como o MAE são utilizaveis em casos aonde temos uma versão controle preenchida dos dados para comparar e quando os ausentes não conterem significado inferivel, como no exemplo da sessão 2 onde removemos aleatoriamente um sub-conjunto dos atributos mas mantivemos um versão controle pra comparação. Alem disso, mesmo que o MAE seja baixo para uma imputação, se houver colisão logica entre o valor imputado e os dados das outras variaveis, esse metodo de avaliação perde rapidamente sua eficacia, não sendo robusto para lidar com as nuances dos dados.

#### Pergunta 2: por que o StandardScaler não resolveu a assimetria?

Na seção 4.1, os três escaladores produziram a mesma assimetria do aluguel original. Por que isso acontece?

### Resposta :
O StandardScaler é um escalonador linear, incapaz de mudar a geometria da distribuição. Ele só consegue mudar a régua dos valores do eixo horizontal. Isso é verdade para todos os métodos lineares apresentados na seção 4.1 do material, onde cada método define uma nova origem e uma nova unidade de medida, mas mantém a forma e a assimetria da distribuição original.

#### Pergunta 3: qual escala você apresentaria ao gestor?

O log reduziu fortemente a assimetria do aluguel. Você apresentaria ao gestor um gráfico do aluguel em escala log?

### Resposta :
Não, eu apresentaria a distribuição de aluguéis em sua escala original, em reais. Embora a escala log aproxime a distribuição de uma normal, reduza a assimetria e estabilize a variância, para um gestor a perda de interpretabilidade causada pela mudança de escala tem um peso muito maior, pois dificulta a compreensão dos valores reais dos aluguéis. A escala original facilita a comunicação dos resultados, a comparação entre metricas calculaveis e a tomada de decisões.

#### 2.Impute o aluguel por tier e zona. Depois, use o bairro como grupo e discuta o efeito de grupos pequenos.

```python
mediana_por_tier = imoveis_missing.groupby("tier").agg(mediana = ("aluguel_mensal","median"))
mediana_por_tier_da_linha = imoveis_missing["tier"].map(mediana_por_tier["mediana"])


mediana_por_bairro = imoveis_missing.groupby("bairro").agg(mediana= ("aluguel_mensal","median"))
mediana_por_bairro_da_linha = imoveis_missing["bairro"].map(mediana_por_bairro["mediana"])

mediana_por_zona = imoveis_missing.groupby("zona").agg(mediana= ("aluguel_mensal","median"))
mediana_por_zona_da_linha = imoveis_missing["zona"].map(mediana_por_zona["mediana"])

imoveis_input_por_tier = imoveis_missing.assign(aluguel_mensal=imoveis_missing["aluguel_mensal"].fillna(mediana_por_tier_da_linha))
imoveis_input_por_zona = imoveis_missing.assign(aluguel_mensal=imoveis_missing["aluguel_mensal"].fillna(mediana_por_zona_da_linha))
imoveis_input_por_bairro = imoveis_missing.assign(aluguel_mensal=imoveis_missing["aluguel_mensal"].fillna(mediana_por_bairro_da_linha))

algueis_input_types = {
    "original": imoveis_true.aluguel_mensal,
    "por_tier": imoveis_input_por_tier.aluguel_mensal,
    "por_zona": imoveis_input_por_zona.aluguel_mensal,
    "por_bairro": imoveis_input_por_bairro.aluguel_mensal,
}

limit = imoveis_true.aluguel_mensal.quantile(.99)

fig, axes = plt.subplots(1, 4, figsize=(20, 4), sharex=True, sharey=True)

for ax, (label, values) in zip(axes, algueis_input_types.items()):
    sns.histplot(
        values,
        bins=50,
        binrange=(0, limit),
        ax=ax,
    )
    ax.set(
        title=label,
        xlabel="aluguel mensal",
        ylabel="Imóveis",
    )
    ax.axvline(values.median(), color="#C44E52", linestyle="--")

plt.tight_layout()
plt.show()

```

Quanto menor o grupo utilizado na imputação, mais a distribuição com os dados imputados se aproximou da distribuição original. Grupos maiores concentram em um único valor uma grande parte dos dados que, normalmente, estariam dispersos entre diferentes valores. Esse valor, a mediana do grupo, acaba sendo muito maior do que o aluguel de uma parte significativa dos imóveis daquele tier, o que distorce a distribuição e cria polos que alteram a moda e o formato da propria distribuição.

#### 3. Entre apartamentos, compare log1p, Box-Cox e Yeo-Johnson para hoa. Qual produz menor assimetria? Qual você apresentaria a um gestor?

```python
imoveis_apartamentos = imoveis_true[imoveis_true["tipo_imovel"]!= "casa"]
condominio = imoveis_apartamentos["condominio_mensal"].dropna()

log = np.log1p(condominio)
boxcox, lambda_bc = stats.boxcox(condominio + 1)
yeojohnson, lambda_yj = stats.yeojohnson(condominio)

print(f"lambda Box-Cox: {lambda_bc:.3f} , lambda Yeo-Johnson: {lambda_yj:.3f}")

assimetria = pd.Series({
    "Original": stats.skew(condominio),
    "log1p": stats.skew(log),
    "Box-Cox": stats.skew(boxcox),
    "Yeo-Johnson": stats.skew(yeojohnson),
})
print(assimetria)
```

```python
versions = {"Original": condominio, "log1p": log, "Box-Cox": boxcox, "Yeo-Johnson": yeojohnson}

fig, axes = plt.subplots(1, 4, figsize=(20, 4))
for ax, (label, v) in zip(axes, versions.items()):
    sns.histplot(v, bins=50, ax=ax)
    ax.set(title=f"{label} (assimetria {stats.skew(v):.2f})", xlabel="Condominio",ylabel="Imóveis")
plt.tight_layout()
plt.show()
```

Apresentaria ao gestor a distribuição original por ser de fácil interpretação e refletir a escala real dos dados. No entanto, a transformação log1p pode ser mais adequada em alguns casos a depender do gestor, pois pode ser convertida facilmente de volta para a escala original e reduz a assimetria o bastante para se equiparar às demais. Já Box-Cox e Yeo-Johnson, embora removam toda a assimetria, dependem de um lambda, o que dificulta a explicação para quem não é da área e torna a reversão menos intuitiva.
